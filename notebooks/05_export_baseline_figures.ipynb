# 05 — Export Baseline Results

**Amaç:** Baseline notebook'unda gerçek olarak üretilen `baseline_history.json` verisini kullanarak ödevde istenen Training/Test Loss ve Accuracy grafiklerini `figures/` klasörüne kaydetmek.

Bu notebook yeni veri üretmez; yalnızca daha önce çalıştırılmış baseline deneyinin kayıtlarını görselleştirir.

```python
# Sonuç dosyasını ve grafik klasörünü kullanmak için temel kütüphaneleri içe aktarırız.
import os
import json
import matplotlib.pyplot as plt

RESULTS = '../results'
FIGURES = '../figures'
os.makedirs(FIGURES, exist_ok=True)

with open(os.path.join(RESULTS, 'baseline_history.json'), 'r', encoding='utf-8') as f:
    history = json.load(f)

epochs = range(1, len(history['train_loss']) + 1)
print('Epoch sayısı:', len(history['train_loss']))
print('Final test accuracy:', history['test_accuracy'][-1])
```

```python
# Training ve test loss eğrisini gerçek baseline kayıtlarından üretir ve SVG olarak kaydederiz.
plt.figure(figsize=(7, 4))
plt.plot(epochs, history['train_loss'], marker='o', label='Training Loss')
plt.plot(epochs, history['test_loss'], marker='o', label='Test Loss')
plt.xlabel('Epoch')
plt.ylabel('Cross-Entropy Loss')
plt.title('Baseline Training vs Test Loss')
plt.legend()
plt.grid(True, alpha=0.25)
plt.tight_layout()
path = os.path.join(FIGURES, 'baseline_loss.svg')
plt.savefig(path, format='svg', bbox_inches='tight')
plt.show()
print('Kaydedildi:', path)
```

```python
# Training ve test accuracy eğrisini gerçek baseline kayıtlarından üretir ve SVG olarak kaydederiz.
plt.figure(figsize=(7, 4))
plt.plot(epochs, history['train_accuracy'], marker='o', label='Training Accuracy')
plt.plot(epochs, history['test_accuracy'], marker='o', label='Test Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('Baseline Training vs Test Accuracy')
plt.legend()
plt.grid(True, alpha=0.25)
plt.tight_layout()
path = os.path.join(FIGURES, 'baseline_accuracy.svg')
plt.savefig(path, format='svg', bbox_inches='tight')
plt.show()
print('Kaydedildi:', path)
```
