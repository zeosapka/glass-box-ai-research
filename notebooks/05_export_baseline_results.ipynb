# 05 — Temel Model Sonuçlarını Dışa Aktarma (Export Baseline Results)

**Amaç:** Temel model notebook'unda gerçek olarak üretilen `baseline_history.json` verisini kullanarak ödevde istenen Eğitim/Test Kaybı (Training/Test Loss) ve Doğruluk (Accuracy) grafiklerini `figures/` klasörüne kaydetmek.

Bu notebook yeni veri üretmez; yalnızca daha önce çalıştırılmış temel model deneyinin kayıtlarını görselleştirir.

```python
# Sonuç dosyasını ve grafik klasörünü kullanmak için temel kütüphaneleri içe aktarırız.
import os
import json
import matplotlib.pyplot as plt

RESULTS = '../results'
FIGURES = '../figures'
os.makedirs(FIGURES, exist_ok=True)

with open(os.path.join(RESULTS, 'baseline_history.json'), 'r', encoding='utf-8') as f:
    history = json.load(f)

epochs = range(1, len(history['train_loss']) + 1)
print('Epoch sayısı:', len(history['train_loss']))
print('Final test accuracy:', history['test_accuracy'][-1])
```

```python
# Eğitim ve test kaybı eğrisini gerçek temel model kayıtlarından üretir ve SVG olarak kaydederiz.
plt.figure(figsize=(7, 4))
plt.plot(epochs, history['train_loss'], marker='o', label='Eğitim Kaybı (Training Loss)')
plt.plot(epochs, history['test_loss'], marker='o', label='Test Kaybı (Test Loss)')
plt.xlabel('Epoch')
plt.ylabel('Çapraz Entropi Kaybı (Cross-Entropy Loss)')
plt.title('Temel Model: Eğitim ve Test Kaybı')
plt.legend()
plt.grid(True, alpha=0.25)
plt.tight_layout()
path = os.path.join(FIGURES, 'baseline_loss.svg')
plt.savefig(path, format='svg', bbox_inches='tight')
plt.show()
print('Kaydedildi:', path)
```

```python
# Eğitim ve test doğruluğu eğrisini gerçek temel model kayıtlarından üretir ve SVG olarak kaydederiz.
plt.figure(figsize=(7, 4))
plt.plot(epochs, history['train_accuracy'], marker='o', label='Eğitim Doğruluğu (Training Accuracy)')
plt.plot(epochs, history['test_accuracy'], marker='o', label='Test Doğruluğu (Test Accuracy)')
plt.xlabel('Epoch')
plt.ylabel('Doğruluk (Accuracy)')
plt.title('Temel Model: Eğitim ve Test Doğruluğu')
plt.legend()
plt.grid(True, alpha=0.25)
plt.tight_layout()
path = os.path.join(FIGURES, 'baseline_accuracy.svg')
plt.savefig(path, format='svg', bbox_inches='tight')
plt.show()
print('Kaydedildi:', path)
```
